```python
# perhitungan faktorial 
n = int(input("masukkan angka: "))
faktorial = 1
i = 1

while i <= n:
    faktorial *= i
    i += 1
print("faktorial dari", n, "adalah", faktorial)
```

```python
# perhitungan jumlah digit
angka = int(input("masukkan angka"))
count = 0

while angka != 0:
    angka //= 10
    count += 1

print("jumlah digit", count)
```

```python
# meghitung rata-rata dengan while
total = 0
count = 0
nilai = float(input("masukkan nilai(-1 untuk berhenti):"))

while nilai != -1:
    total += nilai
    count += 1
    nilai = float(input("masukkan nilai (-1 untuk berhenti): "))

if count > 0:
    rata_rata = total / count
    print("rata-rata nilai:", rata_rata)
    
    print("tidak ada nilai")
```

```python
n = 5
i = 1


#  bagian atas berlian
while i <= n:
    print(" " * (n-i) = n "*" * (2 * i -1)
    i += 1
# bagian bawah berlian
i = n - 1
while i >= 1:
    print(" " * (n - i) + "*" * (2 * i - 1))
    i -= 1
```
