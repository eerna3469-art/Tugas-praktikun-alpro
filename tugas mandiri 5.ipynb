```python
hari = 1
saldo = 50000

while hari <= 30:
   pengeluaran = int(input(f"Pengeluaran hari ke-{hari}: "))
   saldo -= pengeluaran
       
   if saldo < 0:
        print(f"Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break
   print(f"Sisa saldo di hari ke-{hari}: Rp{saldo}")
   hari +=  1

else:
   print(f"catatan pengeluaran hari ke 30 ")
```

```python
pengeluaran_mingguan = [20000, 0, 15000, 0]
pengeluaran = 0

for i in range(4):
    pengeluaran = pengeluaran_mingguan[i]
    minggu_ke = i + 1
    
    if pengeluaran == 0:
        print(f"minggu ke-{minggu_ke}: nobita tidak jajan, delewati analisisnya")
        continue
    print(f"minggu ke-{minggu_ke}: total pengeluaran rp{pengeluaran}")    
    
else:
    print("Analisis mingguan selsai")
```

```python
total_3_bulan = 0

for bulan in range(1, 4):
    total_bulanan = 0
    print(f"Bulan ke-{bulan}")

    for minggu in range(1, 5):
        pengeluaran = int(input(f"pengeluaran minggu ke-{minggu}: Rp"))

        if pengeluaran > 20000:
            print("pengeluaran terlalu besar! Pencatatan bulan ini dihentikan.")
            break

        total_bulanan += pengeluaran
    
    else:
        total_3_bulan += total_bulanan
        print(f"total pengeluaran bulan ke-{bulan}: rp{total_bulanan}")
```

```python
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(11, 13):
    print(f"=== bulan ke-{bulan} ===")

    minggu = 1
    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ")
        
        if libur == "y":
            print("minggu ini di lewati(continue)")
            minggu += 1
            continue
            
        elif libur == "n":
            pengeluaran = int(input("Pengeluaran minggu ini"))
            saldo -= pengeluaran
            minggu += 1

            
        if saldo < 0:
            break
    else:
        print(f"saldo tersisa akhir bulan {bulan}: Rp{saldo}")
        print()
        continue

        break
else:
    print(f"Nobita berhasil menabung RP{saldo} dari total {saldo_awal}!")
         
```
